```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset("iris")
```

```python
#scatter_plot
plt.figure(figsize=(6,4))
sns.scatterplot(
    x="sepal_length",
    y="sepal_width",
    hue="species",
    data=df
)
plt.title("My scatter plot")
plt.show()
```

```python
df.hist(figsize=(6,4))
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(data=df.drop("species", axis=1))
plt.title("My boxplot")
plt.show()
```
